# Clone tree (Newick)
Loads the bifurcating Newick tree produced by the simulation and displays it.

```python
from pathlib import Path
import matplotlib.pyplot as plt
from Bio import Phylo
```

```python
newick_path = Path("out/clone_tree.new")
tree = Phylo.read(newick_path, "newick")
print(f"Loaded {tree.count_terminals()} tips from {newick_path}")
```

```python
fig, ax = plt.subplots(figsize=(12, max(6, tree.count_terminals() * 0.22)))
# Only label tips; internal nodes (incl. the ones added to make the tree bifurcating) stay blank.
Phylo.draw(tree, axes=ax, do_show=False, label_func=lambda c: c.name if c.is_terminal() else "")
ax.set_title(newick_path.name)
plt.tight_layout()
plt.show()
```
