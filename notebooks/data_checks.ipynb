```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
dfs = []
dfs.append(pd.read_csv('../out/22_10_23_10_49_24/summary.csv'))
dfs.append(pd.read_csv('../out/22_10_23_10_57_51/summary.csv'))
dfs.append(pd.read_csv('../out/22_10_23_11_20_32/summary.csv'))


confs = [0, 0.1, 1]
for i in range(len(dfs)):
    dfs[i] = dfs[i][dfs[i]["GenerationId"] == 20]
dfs[0].head()
```

```python
for i in range(len(dfs)):
    print(f'Conf: {confs[i]:.2f}, mean diversity {dfs[i]["ClonalDiversity"].mean():.4f}')
```

```python
plt.gca().xaxis.set_major_formatter(plt.ScalarFormatter())
plt.gca().yaxis.set_major_formatter(plt.ScalarFormatter())
plt.ticklabel_format(useOffset=False)
# set size of plot to 16x9
plt.rcParams["figure.figsize"] = (16,9)

for df in dfs:
    plt.scatter(df["MeanDriversPerCell"], df["ClonalDiversity"])
# apply function to confs
legend = []
for i in range(len(dfs)):
    legend.append(f'Conf: {confs[i]:.1f}, diversity {dfs[i]["ClonalDiversity"].mean():.3f}')
# set legend to confs
plt.legend(legend)
plt.title("Clonal Diversity/Mean Drivers Per Cell, BASIC")
# set both x and y to log scale
plt.ylabel("Clonal Diversity")
plt.xlabel("Mean Drivers Per Cell")
plt.xscale('log')
plt.yscale('log')

```
